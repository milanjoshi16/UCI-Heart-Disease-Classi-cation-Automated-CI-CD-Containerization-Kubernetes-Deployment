```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")

plt.figure(figsize=(6, 4))
ax = sns.countplot(x='target', data=hrt, palette='Set2')
plt.title('Class Balance: Heart Disease Target Distribution', fontsize=12, fontweight='bold')
plt.xlabel('Target (0 = No Disease, 1 = Disease)', fontsize=10)
plt.ylabel('Count', fontsize=10)

# Annotate counts on bars
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 5), textcoords='offset points')

plt.tight_layout()
plt.show()

# Print percentage distribution
print(hrt['target'].value_counts(normalize=True) * 100)
```

# Feature Distribution Histograms

```python
# Plot histograms for key numerical features
numerical_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']

hrt[numerical_cols].hist(bins=15, figsize=(12, 8), color='skyblue', edgecolor='black', grid=False)
plt.suptitle('Histograms of Numerical Features', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

# Feature Correlation Heatmap

```python
import numpy as np

plt.figure(figsize=(12, 10))
correlation_matrix = hrt.corr()

mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            mask=mask, linewidths=0.5, cbar_kws={'shrink': 0.8})

plt.title('Feature Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

# Target-wise Feature Comparison

```python
# Compare average feature values for patients with vs. without heart disease
numeric_features = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
target_comparison = hrt.groupby('target')[numeric_features].mean()
print("--- Average Feature Values by Target Class ---")
display(target_comparison)
```

# Categorical Feature Distributions

```python
# Explore categorical/discrete features against the target
cat_cols = ['sex', 'cp', 'fbs', 'restecg', 'exang', 'slope', 'ca', 'thal']

plt.figure(figsize=(15, 12))
for i, col in enumerate(cat_cols, 1):
    plt.subplot(3, 3, i)
    sns.countplot(x=col, hue='target', data=hrt, palette='Set1')
    plt.title(f'{col} vs Target')
    plt.legend(title='Target', loc='upper right', labels=['No Disease', 'Disease'])
plt.tight_layout()
plt.show()
```

# Outlier Detection using Boxplots

```python
# Visualize outliers in continuous variables using boxplots
plt.figure(figsize=(12, 6))
sns.boxplot(data=hrt[['age', 'trestbps', 'chol', 'thalach', 'oldpeak']], palette='Set3')
plt.title('Boxplots for Outlier Detection in Numerical Features', fontsize=12, fontweight='bold')
plt.ylabel('Values')
plt.show()
```
